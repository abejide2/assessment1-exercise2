12. In the string <code>s3</code>, find the four consicutive digit character using <code>\d</code> and <code>search() </code>function:

```python
s3 = "House number- 1105"

# Write your code below and press Shift+Enter to execute
```

```python
import re

# Given string
s3 = "House number- 1105"

# Search for exactly four consecutive digits
match = re.search(r"\d{4}", s3)

if match:
    print(f"Found: {match.group()} at position {match.start()}–{match.end()-1}")
else:
    print("No four consecutive digits found.")
```
